# Treinamento dos seis experimentos

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.config import EXPERIMENTOS, BATCH_SIZE, EPOCHS, PACIENCIA, LEARNING_RATE, SEED

print("semente", SEED)
print("lote", BATCH_SIZE, "| épocas máximas", EPOCHS, "| paciência", PACIENCIA)
print("Adam lr", LEARNING_RATE, "| perda sparse_categorical_crossentropy")
display(pd.DataFrame(EXPERIMENTOS))
```
