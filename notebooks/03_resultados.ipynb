```python
erros = resumo["erros"]
print(
    f"{erros['experimento']}: {erros['total_erros_teste']} erros "
    f"em {erros['total_teste']} imagens de teste."
)
display(pd.DataFrame(erros["amostras"]))
```

```python
import numpy as np

# Erros e acertos do modelo escolhido abaixo de um limite de probabilidade da classe prevista (p).
previsoes = np.load(ROOT / "artifacts" / "metricas" / "previsoes.npz")
real = previsoes["y_teste"]
previsto = previsoes[f"{erros['experimento']}_teste"]
p = previsoes[f"{erros['experimento']}_proba_teste"][np.arange(len(real)), previsto]
errou = previsto != real
limite = 0.85
abaixo = p < limite
print(f"Previsões com p abaixo de {limite}: {int(abaixo.sum())}, sendo {int((errou & abaixo).sum())} erros e {int((~errou & abaixo).sum())} acertos.")
print(f"Erros com p de {limite} ou mais: {int((errou & ~abaixo).sum())} de {int(errou.sum())}.")
```

## Análise

- Arquiteturas: em 128 × 96, a CNN teve F1 macro 0,657 contra 0,342 da MLP, com 46 vezes menos parâmetros.
- Desequilíbrio: os pesos de classe levaram o recall de trash de 0,20 para 0,55 na CNN, com F1 macro quase igual, e na MLP baixaram o F1 macro de 0,342 para 0,309.
- Resolução e custo: a 256 × 192 subiu o F1 macro de 0,657 para 0,718 com 15,7 minutos de treino, e a 512 × 384 levou 62,6 minutos e 3.291 MB de memória e caiu para 0,685.
- Erros: o modelo escolhido (CNN 256 × 192) errou 93 das 380 imagens de teste, 41 delas em trocas entre glass, plastic e metal, e, treinado com um objeto por foto sobre fundo branco, não poderia ser usado sem novos testes numa esteira de reciclagem.
- Yang e Thung: o artigo deles teve 63% de acurácia com SVM em seis classes e 22% com uma CNN em cinco classes, e a CNN 256 × 192 deste trabalho teve 75,5% em seis, mas classes, divisão, resolução e treino diferem, então os números não se comparam diretamente.

O modelo da figura de erros foi escolhido pelo maior F1 macro na validação; o teste não foi usado na escolha.
